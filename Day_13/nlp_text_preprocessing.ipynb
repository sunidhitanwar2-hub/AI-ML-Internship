```python
import pandas as pd
import numpy as np
import re

from sklearn.feature_extraction.text import TfidfVectorizer

print("Libraries imported successfully!")
```

```python
texts = [
    "Machine Learning is an interesting field.",
    "Python is useful for Machine Learning.",
    "Natural Language Processing helps computers understand text.",
    "I am learning Python and Artificial Intelligence.",
    "Text preprocessing is important in NLP."
]

df = pd.DataFrame({"Text": texts})

display(df)
```

```python
df["Lowercase_Text"] = df["Text"].str.lower()

display(df)
```

```python
df["Tokens"] = df["Lowercase_Text"].apply(lambda text: text.split())

display(df[["Lowercase_Text", "Tokens"]])
```

```python
stop_words = {
    "is", "an", "a", "the", "and", "in",
    "for", "are", "to", "of", "i", "am"
}

df["Filtered_Tokens"] = df["Tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)

display(df[["Tokens", "Filtered_Tokens"]])
```

```python
# Convert filtered tokens back into sentences
cleaned_texts = df["Filtered_Tokens"].apply(
    lambda tokens: " ".join(tokens)
)

# Create TF-IDF vectorizer
tfidf = TfidfVectorizer()

# Convert text into numerical values
tfidf_matrix = tfidf.fit_transform(cleaned_texts)

print("TF-IDF vectorization completed successfully!")
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

display(tfidf_df.round(3))
```

```python
tfidf_df = pd.DataFrame(
    tfidf_matrix.toarray(),
    columns=tfidf.get_feature_names_out()
)

display(tfidf_df.round(3))
```

```python
print("Unique words in the dataset:")
print(tfidf.get_feature_names_out())

print("\nTF-IDF Matrix Shape:")
print(tfidf_matrix.shape)
```
