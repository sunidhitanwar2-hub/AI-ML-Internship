```python
import numpy as np

arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("1D Array:", arr1)
print("2D Array:\n", arr2)
```

```python
print("First element:", arr1[0])
print("Last element:", arr1[-1])
print("First three elements:", arr1[:3])
print("First row of 2D array:", arr2[0])
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
print("Multiply by 2:", numbers * 2)
```

```python
python_list = [10, 20, 30]
numpy_array = np.array([10, 20, 30])

print("Python list:", python_list * 2)
print("NumPy array:", numpy_array * 2)
```
