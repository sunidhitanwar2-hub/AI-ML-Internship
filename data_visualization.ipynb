### Charts

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("online_sales.csv")

sns.set_theme(style="whitegrid")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python

# Calculate revenue for each order
df["Revenue"] = df["Price"] * df["Quantity"]

# Calculate total revenue by category
category_revenue = df.groupby("Category")["Revenue"].sum()
category_revenue = category_revenue.sort_values(ascending=False)

# Create the bar chart
plt.figure(figsize=(8, 5))

sns.barplot(
    x=category_revenue.index,
    y=category_revenue.values,
    color="steelblue"
)

plt.title("Total Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Line chart: Revenue by order

plt.figure(figsize=(9, 5))

plt.plot(
    range(1, len(df) + 1),
    df["Revenue"],
    marker="o",
    linewidth=2
)

plt.title("Revenue by Order")
plt.xlabel("Order Number")
plt.ylabel("Revenue (₹)")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Histogram: Distribution of order revenue

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Revenue",
    bins=8,
    kde=True,
    color="teal"
)

plt.title("Distribution of Order Revenue")
plt.xlabel("Revenue per Order (₹)")
plt.ylabel("Number of Orders")
plt.tight_layout()
plt.show()
```

```python

# Pie chart: Revenue share by category

plt.figure(figsize=(7, 7))

category_revenue.plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Revenue Share by Product Category")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

```python

print("KEY OBSERVATIONS")
print("-" * 40)

print(
    f"1. Highest revenue category: {category_revenue.idxmax()} "
    f"(₹{category_revenue.max():,.2f})"
)

print(
    f"2. Lowest revenue category: {category_revenue.idxmin()} "
    f"(₹{category_revenue.min():,.2f})"
)

print(
    f"3. Average revenue per order: ₹{df['Revenue'].mean():,.2f}"
)

print(
    f"4. Highest individual order revenue: ₹{df['Revenue'].max():,.2f}"
)

print(
    f"5. Total revenue across all orders: ₹{df['Revenue'].sum():,.2f}"
)
```
